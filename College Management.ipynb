## 📊 ROUND 2 — STATISTICS BATTLE

### Mission: Describe the typical employee salary and its variation.

```python
mean_cgpa = df["CGPA"].mean()

median_cgpa = df["CGPA"].median()

mode_cgpa = df["CGPA"].mode()[0]

variance_cgpa = df["CGPA"].var()

std_cgpa = df["CGPA"].std()

cv_cgpa = (std_cgpa / mean_cgpa) * 100

# Display results
print(f"Mean CGPA: {mean_cgpa:.2f}")
print(f"Median CGPA: {median_cgpa:.2f}")
print(f"Mode CGPA: {mode_cgpa:.2f}")
print(f"Variance: {variance_cgpa:.2f}")
print(f"Standard Deviation: {std_cgpa:.2f}")
print(f"Coefficient of Variation: {cv_cgpa:.2f}%")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df['CGPA'], kde=True, bins=20, color='skyblue')
plt.title('Distribution of CGPA')
plt.xlabel('CGPA')
plt.ylabel('Frequency / Density')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
Q1 = df["CGPA"].quantile(0.25)
Q3 = df["CGPA"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["CGPA"] < lower) |
    (df["CGPA"] > upper)
]

print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower Bound: {lower:.2f}")
print(f"Upper Bound: {upper:.2f}")
print(f"Outliers Found: {len(outliers)}")

display(outliers.head(10))
```

# 🎲 ROUND 6 — PROBABILITY ***ROULETTE***

```python
df["Placement_Flag"] = (
    df["Placement_Status"] == "Selected"
).astype(int)

# Overall probability of placement
overall_placement_rate = df["Placement_Flag"].mean()

# Conditional probability: P(Placement | CGPA >= 8)
high_cgpa = df[
    df["CGPA"] >= 8
]

conditional_placement_rate = (
    high_cgpa["Placement_Flag"].mean()
)

# Display results
print(
    f"Overall Placement Rate: "
    f"{overall_placement_rate:.2%}"
)

print(
    "P(Placement | CGPA >= 8): "
    f"{conditional_placement_rate:.2%}"
)
```

🧪 ROUND 7 — HYPOTHESIS TEST ARENA

```python
wssssssssfrom scipy import stats

selected = df[
    df["Placement_Status"] == "Selected"
]["CGPA"]

not_selected = df[
    df["Placement_Status"] == "Not Selected"
]["CGPA"]

t_stat, p_value = stats.ttest_ind(
    selected,
    not_selected,
    equal_var=False
)

print(f"t-statistic: {t_stat:.3f}")
print(f"p-value: {p_value:.4f}")

if p_value < 0.05:
    print("Decision: Reject H0 at the 5% significance level.")
else:
    print("Decision: Insufficient evidence to reject H0 at the 5% significance level.")
```

Effect Size Challenge — Cohen's **d**

```python
import numpy as np

pooled_std = np.sqrt(
    (selected.var() + not_selected.var()) / 2
)

cohens_d = (
    selected.mean() - not_selected.mean()
) / pooled_std

print(f"Cohen's d: {cohens_d:.2f}")
```
